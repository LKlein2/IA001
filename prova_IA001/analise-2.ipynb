### Tratamento dos dados (usado pelos três gráficos e pelo `dashboard.py`)

- **Zona Sul:** a coluna `neighbourhood_group` está 100% vazia, então a região é definida por uma lista fixa — os 18 bairros da Área de Planejamento 2.1 da Prefeitura do Rio.
- **Anunciante sem identificação:** 14 anúncios sem `host_id` são removidos (não dá para classificá-los).
- **Categoria do anunciante** (a partir de `calculated_host_listings_count`, que conta os anúncios do anfitrião **na cidade toda**):
  - *Profissional* = imóvel inteiro (`Entire home/apt`) **e** anfitrião com mais de 1 anúncio;
  - *Individual* = anfitrião com 1 único anúncio;
  - *Outros* = anfitrião com vários anúncios, mas oferecendo quarto (privativo, compartilhado ou de hotel). Não cabe em nenhuma das duas definições da pergunta; fica como categoria explícita em vez de ser descartado.
- **Preço** não entra nas perguntas → os 4.172 preços ausentes e os extremos (máx. R$ 574 mil) não afetam a análise e não foram tratados.
- A função `preparar` é **idêntica** à do `dashboard.py`, para que os números do notebook e do app coincidam.

````python
ZONA_SUL = [
    "Botafogo", "Catete", "Copacabana", "Cosme Velho", "Flamengo", "Gávea",
    "Glória", "Humaitá", "Ipanema", "Jardim Botânico", "Lagoa", "Laranjeiras",
    "Leblon", "Leme", "Rocinha", "São Conrado", "Urca", "Vidigal",
]
PROFISSIONAL, INDIVIDUAL, OUTROS = "Profissional", "Individual", "Outros"
CATEGORIAS = [PROFISSIONAL, INDIVIDUAL, OUTROS]
CORES = {PROFISSIONAL: "#2a78d6", INDIVIDUAL: "#eb6834", OUTROS: "#9a9890"}


def preparar(df):
    """Recorta a Zona Sul e classifica cada anúncio por categoria de anunciante."""
    df = df.dropna(subset=["host_id"])
    zs = df[df["neighbourhood"].isin(ZONA_SUL)].copy()
    inteiro = zs["room_type"] == "Entire home/apt"
    varios = zs["calculated_host_listings_count"] > 1
    zs["categoria"] = np.select(
        [inteiro & varios, ~varios],
        [PROFISSIONAL, INDIVIDUAL],
        default=OUTROS,
    )
    zs["categoria"] = pd.Categorical(zs["categoria"], categories=CATEGORIAS)
    return zs


zs = preparar(df)
cidade = df.dropna(subset=["host_id"])
print(f"Bairros da lista encontrados no arquivo: {zs['neighbourhood'].nunique()} de {len(ZONA_SUL)}")
print(f"Anúncios na Zona Sul: {len(zs):,} ({len(zs) / len(cidade):.1%} da cidade)")
print(zs["categoria"].value_counts().to_frame("anúncios").assign(
    pct=lambda t: (t["anúncios"] / t["anúncios"].sum() * 100).round(1)))
````

### Gráfico 1 — Anúncios por bairro da Zona Sul (P1)

> ✏️ *Rascunho em tópicos — reescrever com suas palavras.*

**Justificativa do tipo:**
- Comparar magnitudes entre categorias nominais (bairros) → barras; horizontais porque os nomes são longos e ficam legíveis sem rotação.
- Ordenadas do maior para o menor; o rótulo traz o **% acumulado**, que mostra a concentração (lógica de Pareto) sem criar um segundo eixo.
- Uma só série → sem legenda (o título a nomeia).

**Decisões de tratamento:**
- Contagem bruta de anúncios, todos os 18 bairros, sem filtro.
- Escala **linear**: a escala log "achataria" a diferença de Copacabana para os demais, que é justamente o achado.
- Não normalizado pela área nem pelo número de domicílios do bairro (ver limitações).

````python
contagem = zs["neighbourhood"].value_counts()
ranking = pd.DataFrame({"bairro": contagem.index, "n": contagem.values})
ranking["pct"] = ranking["n"] / ranking["n"].sum()
ranking["pct_acum"] = ranking["pct"].cumsum()
ranking["rotulo"] = [f"{n:,} ({p:.1%} acum.)".replace(",", "X").replace(".", ",").replace("X", ".")
                     for n, p in zip(ranking["n"], ranking["pct_acum"])]

fig1 = px.bar(
    ranking, x="n", y="bairro", orientation="h", text="rotulo",
    title="Anúncios por bairro da Zona Sul",
    labels={"n": "Número de anúncios", "bairro": "Bairro"},
)
fig1.update_traces(marker_color="#2a78d6", textposition="outside", cliponaxis=False)
fig1.update_layout(
    yaxis={"categoryorder": "total ascending", "title_standoff": 12, "automargin": True},
    height=620, margin={"l": 150, "r": 130}, separators=",.",
)
fig1.show()

n80 = int((ranking["pct_acum"] < 0.8).sum()) + 1
print(f"{n80} de {len(ranking)} bairros somam 80% da oferta da Zona Sul")
ranking[["bairro", "n", "pct", "pct_acum"]].head(8).round(3)
````

### Gráfico 3 — Composição da oferta por categoria de anunciante nos 5 bairros com mais anúncios (P2)

> ✏️ *Rascunho em tópicos — reescrever com suas palavras.*

**Justificativa do tipo:**
- A pergunta é sobre **proporção dentro de cada bairro** → barras 100% empilhadas: todas as barras têm o mesmo comprimento e o que se compara é a divisão interna.
- O n de anúncios entra no rótulo do bairro, para não esconder que Copacabana (15.036) é 15 vezes maior que Flamengo (1.032).
- Ordenadas pelo % profissional; legenda com as três categorias e cores iguais às do mapa.

**Decisões de tratamento:**
- "Bairros com maior oferta" = os **5 maiores**, que somam 83% da oferta da Zona Sul (ver gráfico 1); no dashboard o número é ajustável.
- Bairros pequenos (Cosme Velho 90, Rocinha 93) ficariam com percentuais instáveis — poucos anfitriões mudam muito o resultado.
- A categoria *Outros* é mantida na barra (não redistribuída), para que Profissional + Individual não pareçam somar 100%.

````python
TOP_N = 5
maiores = contagem.head(TOP_N).index
sub = zs[zs["neighbourhood"].isin(maiores)]
composicao = pd.crosstab(sub["neighbourhood"], sub["categoria"], normalize="index")
composicao = composicao.sort_values(PROFISSIONAL)

longo = (
    composicao.rename(index=lambda b: f"{b} ({contagem[b]:,})".replace(",", "."))
    .reset_index().melt(id_vars="neighbourhood", var_name="categoria", value_name="pct")
)
fig3 = px.bar(
    longo, x="pct", y="neighbourhood", color="categoria", orientation="h", text="pct",
    color_discrete_map=CORES, category_orders={"categoria": CATEGORIAS},
    title=f"Composição da oferta por categoria de anunciante nos {TOP_N} bairros com mais anúncios",
    labels={"pct": "Participação na oferta do bairro (%)",
            "neighbourhood": "Bairro (número de anúncios)", "categoria": "Categoria de anunciante"},
)
fig3.update_traces(texttemplate="%{text:.0%}", textposition="inside",
                   marker_line_color="white", marker_line_width=2)
fig3.update_layout(
    barmode="stack", xaxis={"tickformat": ".0%", "range": [0, 1]},
    yaxis={"title_standoff": 12, "automargin": True},
    legend={"orientation": "h", "yanchor": "bottom", "y": 1.0, "x": 0},
    height=400, margin={"l": 190, "t": 110}, separators=",.",
)
fig3.show()

tabela_p2 = pd.crosstab(sub["neighbourhood"], sub["categoria"])
tabela_p2["total"] = tabela_p2.sum(axis=1)
tabela_p2.join((composicao * 100).round(1), rsuffix=" %").sort_values("total", ascending=False)
````

### Reflexão Crítica

> ✏️ *Rascunho em tópicos — reescrever com suas palavras.* *(os números vêm das células abaixo)*

1. **Sensibilidade da definição de "profissional".** Se *Outros* contar como profissional (qualquer anfitrião com 2+ anúncios), a fatia profissional da ZS sobe de 46,5% para 57,6% e **Botafogo vira majoritariamente profissional (32,8% → 50,4%)** — a resposta "Botafogo é individual" depende da definição. Exigindo 3+ ou 11+ imóveis inteiros, a fatia cai (36,1% e 18,9% na ZS), mas Ipanema/Leblon seguem no topo e Botafogo/Flamengo embaixo nos 5 maiores bairros; nos 18 bairros, a correlação de Spearman do ranking com o da definição base é 0,96 (3+) e 0,77 (11+). → A conclusão "orla mais profissional que Botafogo/Flamengo" é robusta; o *nível* (46%, 58%…) não é.
2. **Verificação independente.** Recontei os anúncios por `host_id` (groupby na cidade toda) e comparei com `calculated_host_listings_count`: bateu em 100% dos registros → a variável que define a categoria é consistente com os próprios dados. Também conferi pelas coordenadas que 100% dos anúncios rotulados "Copacabana" caem numa caixa geográfica do bairro (verificação fraca: a caixa é retangular e também pega Leme e parte de Ipanema; não confirma a fronteira, só que não há rótulos absurdos).
3. **Anúncios inativos.** Refiz as duas medidas só com anúncios com avaliação nos últimos 12 meses (`number_of_reviews_ltm > 0`). 72,8% dos anúncios da ZS são ativos. Copacabana sobe de 51,9% para 53,7% da oferta, e a fatia profissional **aumenta** nos 5 bairros (ex.: Botafogo 32,8% → 37,2%), porque os inativos são mais frequentes entre individuais (30,1%) e *Outros* (37,5%) do que entre profissionais (22,2%) → a concentração e a profissionalização não são efeito de anúncios abandonados; se algo, estão subestimadas.
4. **Explicação alternativa.** A maior fatia profissional em Ipanema/Leblon pode não ser "investidores comprando para Airbnb", mas prédios de flats/apart-hotéis já existentes cadastrados como imóveis inteiros, ou administradoras que gerenciam imóveis de vários donos distintos.
5. **Qualidade do dado.** `last_review` chega a 30/06/2026, posterior à data de coleta anotada (24/06) — ou a data de coleta está errada, ou o arquivo foi composto em mais de um dia. Não afeta as contagens, mas mostra que metadados precisam ser conferidos.

````python
# 1) Sensibilidade: % "profissional" por bairro sob definições alternativas
base = zs.dropna(subset=["calculated_host_listings_count"])
inteiro = base["room_type"] == "Entire home/apt"
n_anuncios = base["calculated_host_listings_count"]
definicoes = {
    "base: inteiro e 2+": inteiro & (n_anuncios > 1),
    "qualquer tipo e 2+ (Outros→prof.)": n_anuncios > 1,
    "inteiro e 3+": inteiro & (n_anuncios >= 3),
    "inteiro e 11+": inteiro & (n_anuncios >= 11),
}
sens = pd.DataFrame({nome: mascara.groupby(base["neighbourhood"]).mean() for nome, mascara in definicoes.items()})
sens.loc["ZONA SUL (total)"] = [m.mean() for m in definicoes.values()]
sens = sens.loc[list(maiores) + ["ZONA SUL (total)"]]
print("% profissional nos 5 maiores bairros, por definição")
display((sens * 100).round(1))

# A ordem dos bairros muda? Correlação de Spearman entre rankings (18 bairros)
todos = pd.DataFrame({nome: m.groupby(base["neighbourhood"]).mean() for nome, m in definicoes.items()})
print("Spearman entre rankings dos 18 bairros:")
display(todos.rank().corr().round(2))
````

````python
# 2) Verificação independente
recontagem = cidade.groupby("host_id")["id"].transform("size")
igual = (recontagem == cidade["calculated_host_listings_count"]).mean()
print(f"calculated_host_listings_count == recontagem por host_id: {igual:.1%} dos anúncios")

copa = zs[zs["neighbourhood"] == "Copacabana"]
caixa = copa["latitude"].between(-22.995, -22.955) & copa["longitude"].between(-43.200, -43.163)
print(f"Anúncios rotulados 'Copacabana' dentro da caixa geográfica: {caixa.mean():.1%}")
na_caixa = zs[zs["latitude"].between(-22.995, -22.955) & zs["longitude"].between(-43.200, -43.163)]
print("Anúncios da ZS dentro da caixa, por rótulo de bairro (a caixa também pega Leme e parte de Ipanema):")
display(na_caixa["neighbourhood"].value_counts().head(5))
print(f"Copacabana / Zona Sul (contagem direta): {len(copa) / len(zs):.1%}  |  / cidade: {len(copa) / len(cidade):.1%}")

# 3) Só anúncios ativos (com avaliação nos últimos 12 meses)
ativos = zs[zs["number_of_reviews_ltm"] > 0]
print(f"\nAnúncios ativos na ZS: {len(ativos):,} de {len(zs):,} ({len(ativos) / len(zs):.1%})")
comparacao = pd.DataFrame({
    "todos": zs["neighbourhood"].value_counts(normalize=True),
    "ativos": ativos["neighbourhood"].value_counts(normalize=True),
}).loc[list(maiores)]
print("Participação dos 5 maiores bairros na oferta:")
display((comparacao * 100).round(1))
prof_ativos = pd.DataFrame({
    "todos": (zs["categoria"] == PROFISSIONAL).groupby(zs["neighbourhood"]).mean(),
    "ativos": (ativos["categoria"] == PROFISSIONAL).groupby(ativos["neighbourhood"]).mean(),
}).loc[list(maiores)]
print("% profissional nos 5 maiores bairros:")
display((prof_ativos * 100).round(1))
print("Inativos por categoria (% sem avaliação em 12 meses):")
display(((zs["number_of_reviews_ltm"] == 0).groupby(zs["categoria"], observed=True).mean() * 100).round(1))

# 5) Datas
print(f"\nÚltima avaliação mais recente no arquivo: {df['last_review'].max()}")
````
